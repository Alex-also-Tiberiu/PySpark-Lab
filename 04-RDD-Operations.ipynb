```python
# Import SparkSession
from pyspark.sql import SparkSession
```

```python
# Create a SparkSession
spark = SparkSession.builder.appName("RDD-demo").getOrCreate()
```

## How to create RDDs

```python
numbers = [1,2,3,4,5]
rdd = spark.sparkContext.parallelize(numbers)
```

```python
# Collect action: Retrieve all elements of the RDD
rdd.collect()
```

```python
# Create an RDD from a list of tuples
data = [("Alice", 25), ("Bob", 30), ("Charlie", 35), ("Alice", 40)]
rdd = spark.sparkContext.parallelize(data)
```

```python
# Collect action: Retrieve all elements of the RDD
rdd.collect()
```

## RDD Operation: Actions

```python
count = rdd.count()
print("Total:", count) 
```

```python
# First action: Retrieve the first element of the RDD
first_element = rdd.first()
print("RDD first element:", first_element)
```

```python
# Take action: Retrive n elements of the RDD
n_elements = rdd.take(2)
print("RDD first 2 elements:", n_elements)
```

```python
# Foreach action: execute lambda function for each element of the RDD
rdd.foreach(lambda x: print(x))
```

## RDD (Resilient Distributed Dataset) Transformations

```python
# Map transformation: Convert name to uppercase
mapped_rdd = rdd.map(lambda x: (x[0].upper(), x[1]))
```

```python
result = mapped_rdd.collect()
print("RDD with transformation:", result)
```

```python
# Filter transfomration: Filter is like WHERE clause in SQL
filtered_rdd = rdd.filter(lambda x: x[1] > 30)
filtered_rdd.collect()
```

```python
# ReduceByKey transformation: calculate totale age for each name
filtered_rdd = rdd.reduceByKey(lambda x, y: x + y)
filtered_rdd.collect()
```

```python
# SortBy transformation: Sort the RDD by age in descending order
sorted_rdd = rdd.sortBy(lambda x: x[1], ascending=False)
sorted_rdd.collect()
```

## RDD Read and Write from text file

```python
# Save action: Save RDD to text file
rdd.saveAsTextFile("output.txt")
```
