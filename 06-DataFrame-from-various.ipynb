```python
# Import SparkSession and so on
from pyspark.sql import SparkSession
from pyspark.sql.functions import desc
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType
```

```python
# Create a SparkSession
spark = SparkSession.builder.appName("DataFrame-demo").getOrCreate()
```

```python
%%bash
head -10 ./data/products.csv
```

## Read CSV intro DataFrame

```python
df = spark.read.csv("./data/products.csv", header=True)
```

```python
# Display schema
df.printSchema()

# Display content
df.show(5)
```

## Read CSV using StructType

```python
# Define schema with StructType
schema = StructType([
    StructField(name="id",       dataType=IntegerType() ,nullable=True),
    StructField(name="name",     dataType=StringType()  ,nullable=True),
    StructField(name="category", dataType=StringType()  ,nullable=True),
    StructField(name="quantity", dataType=IntegerType() ,nullable=True),
    StructField(name="price",    dataType=DoubleType()  ,nullable=True)
])
```

```python
# Read CSV file into DataFrame with schema definition
df = spark.read.csv("./data/products.csv", header=True, schema=schema)
```

```python
# Display schema
df.printSchema()

# Display content
df.show(5)
```

## Read CSV using inferred schema

```python
# Read CSV file into DataFrame with inferred schema
df = spark.read.csv("./data/products.csv", header=True, inferSchema=True)
```

```python
# Display inferred schema
df.printSchema()

# Display content
df.show(5)
```
